```python
import pandas as pd
import os

METADATA_PATH = os.path.abspath(
    "../data/test_metadata (1).csv"
)

print("Metadata path:")
print(METADATA_PATH)

print()
print("File exists:", os.path.exists(METADATA_PATH))

test_metadata = pd.read_csv(METADATA_PATH)

print()
print("Metadata shape:", test_metadata.shape)

print()
print("Columns:")
print(test_metadata.columns.tolist())

print()
print("Language counts:")
print(test_metadata["Language"].value_counts())
```

```python
import numpy as np
import os

# Project data folder
data_dir = r"C:\Users\dimpl\OneDrive\Desktop\VocaTrust\data"

# File paths
X_test_path = os.path.join(data_dir, "X_test (2).npy")
y_test_path = os.path.join(data_dir, "y_test (1).npy")

# Check files
print("X_test exists:", os.path.exists(X_test_path))
print("y_test exists:", os.path.exists(y_test_path))

# Load files
X_test = np.load(X_test_path)
y_test = np.load(y_test_path)

print()
print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)
```

```python
import numpy as np

# Convert metadata GroundTruth into numeric labels
metadata_labels = np.where(
    test_metadata["GroundTruth"] == "REAL",
    1,
    0
)

print("Metadata labels shape:", metadata_labels.shape)
print("y_test shape:", y_test.shape)

print()
print("Labels identical:",
      np.array_equal(metadata_labels, y_test))

print()
print("Metadata REAL:", np.sum(metadata_labels == 1))
print("Metadata FAKE:", np.sum(metadata_labels == 0))

print()
print("y_test REAL:", np.sum(y_test == 1))
print("y_test FAKE:", np.sum(y_test == 0))
```
