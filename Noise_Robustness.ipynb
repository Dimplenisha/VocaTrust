```python
import os
import random
import soundfile as sf
import pandas as pd

# Dataset paths
real_path = "/content/drive/MyDrive/VocaTrust/clean_data/real"
fake_path = "/content/drive/MyDrive/VocaTrust/clean_data/fake"

# Number of files to check from each folder
SAMPLE_SIZE = 50   # 50 Real + 50 Fake = 100 files

records = []

def check_random_files(folder_path, label):
    files = [f for f in os.listdir(folder_path) if f.lower().endswith(".wav")]

    # If folder has fewer than SAMPLE_SIZE files, use all of them
    selected_files = random.sample(files, min(SAMPLE_SIZE, len(files)))

    for file in selected_files:
        file_path = os.path.join(folder_path, file)

        try:
            info = sf.info(file_path)

            records.append({
                "Label": label,
                "Filename": file,
                "Sample Rate (Hz)": info.samplerate,
                "Sample Rate (kHz)": info.samplerate / 1000,
                "Duration (sec)": round(info.duration, 3)
            })

        except Exception as e:
            print(f"Error reading {file}: {e}")

# Check 50 random files from each class
check_random_files(real_path, "Real")
check_random_files(fake_path, "Fake")

df = pd.DataFrame(records)

print(df.head(10))
```

```python
import os
import random

# Paths
REAL_DIR = "/content/drive/MyDrive/VocaTrust/clean_data/real"
NOISE_DIR = "/content/drive/MyDrive/VocaTrust/Phase4_Noise_Robustness/noise_samples"

# One random voice file
voice_file = random.choice([
    os.path.join(REAL_DIR, f)
    for f in os.listdir(REAL_DIR)
    if f.endswith(".wav")
])

# One random noise file
noise_files = []

for root, _, files in os.walk(NOISE_DIR):
    for f in files:
        if f.endswith(".wav"):
            noise_files.append(os.path.join(root, f))

noise_file = random.choice(noise_files)

print("Voice File:")
print(voice_file)

print("\nNoise File:")
print(noise_file)
```

```python
import librosa

voice, sr = librosa.load(voice_file, sr=16000)
noise, _ = librosa.load(noise_file, sr=16000)

print("Voice length :", len(voice))
print("Noise length :", len(noise))
print("Sample Rate  :", sr)
```

```python
import numpy as np

if len(noise) < len(voice):
    repeats = int(np.ceil(len(voice) / len(noise)))
    noise = np.tile(noise, repeats)

noise = noise[:len(voice)]

print("Adjusted Noise Length:", len(noise))
```

```python
#Create the SNR Mixing Function


import numpy as np

def add_noise_snr(clean, noise, snr_db):
    """
    Add noise to a clean signal at a desired SNR.

    Parameters:
        clean  : Clean speech signal
        noise  : Noise signal (same length as clean)
        snr_db : Desired SNR in dB

    Returns:
        Noisy speech signal
    """

    # Power of clean speech
    clean_power = np.mean(clean ** 2)

    # Power of noise
    noise_power = np.mean(noise ** 2)

    # Required noise power
    desired_noise_power = clean_power / (10 ** (snr_db / 10))

    # Scale the noise
    noise = noise * np.sqrt(desired_noise_power / noise_power)

    # Mix
    noisy = clean + noise

    # Normalize to avoid clipping
    max_amp = np.max(np.abs(noisy))

    if max_amp > 1:
        noisy = noisy / max_amp

    return noisy
```

```python
#Mix One File ,We'll start with 15 dB SNR.

snr = 15

noisy_voice = add_noise_snr(voice, noise, snr)

print("Noisy audio created successfully.")
```

```python
import os
import random
import numpy as np
import pandas as pd
import librosa
import soundfile as sf
from tqdm import tqdm
```

```python
#Define Paths
# Clean dataset
REAL_DIR = "/content/drive/MyDrive/VocaTrust/clean_data/real"

# MUSAN noise
NOISE_DIR = "/content/drive/MyDrive/VocaTrust/Phase4_Noise_Robustness/noise_samples"

# Output
OUTPUT_REAL = "/content/drive/MyDrive/VocaTrust/Phase4_Noise_Robustness/noisy_data/real"

LOG_DIR = "/content/drive/MyDrive/VocaTrust/Phase4_Noise_Robustness/logs"

os.makedirs(OUTPUT_REAL, exist_ok=True)
os.makedirs(LOG_DIR, exist_ok=True)
```

```python
#Read All Noise Files and Shuffle
noise_files = []

for root, _, files in os.walk(NOISE_DIR):
    for file in files:
        if file.endswith(".wav"):
            noise_files.append(os.path.join(root, file))

random.shuffle(noise_files)

print("Total Noise Files:", len(noise_files))
```

```python
#SNR Function

def add_noise_snr(clean, noise, snr_db):

    clean_power = np.mean(clean ** 2)
    noise_power = np.mean(noise ** 2)

    desired_noise_power = clean_power / (10 ** (snr_db / 10))

    noise = noise * np.sqrt(desired_noise_power / noise_power)

    noisy = clean + noise

    max_amp = np.max(np.abs(noisy))

    if max_amp > 1:
        noisy = noisy / max_amp

    return noisy
```

```python
#Process the Real Folder
snr_values = [5, 10, 15, 20]

log = []

noise_index = 0

real_files = sorted([
    f for f in os.listdir(REAL_DIR)
    if f.endswith(".wav")
])

for file in tqdm(real_files):

    input_path = os.path.join(REAL_DIR, file)
    output_path = os.path.join(OUTPUT_REAL, file)

    # Skip if already processed
    if os.path.exists(output_path):
        continue

    # Load clean speech
    clean, sr = librosa.load(input_path, sr=16000)

    # Get next noise file
    noise_path = noise_files[noise_index]

    noise_index += 1

    # Reshuffle after all noise files are used
    if noise_index >= len(noise_files):
        random.shuffle(noise_files)
        noise_index = 0

    # Load noise
    noise, _ = librosa.load(noise_path, sr=16000)

    # Match length
    if len(noise) < len(clean):
        repeats = int(np.ceil(len(clean) / len(noise)))
        noise = np.tile(noise, repeats)

    noise = noise[:len(clean)]

    # Random SNR
    snr = random.choice(snr_values)

    # Add noise
    noisy = add_noise_snr(clean, noise, snr)

    # Save
    sf.write(output_path, noisy, sr)

    # Log
    log.append({
        "Original File": file,
        "Noise File": os.path.basename(noise_path),
        "SNR": snr,
        "Output File": file
    })
```

```python
#Save the Log
log_df = pd.DataFrame(log)

log_df.to_csv(
    os.path.join(LOG_DIR, "real_augmentation_log.csv"),
    index=False
)

print("Completed Successfully!")
```
